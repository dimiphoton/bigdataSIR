```python
import arviz as az
import pandas as pd
import pymc as pm
import matplotlib.pyplot as plt

from seir_model.model import build_model
```

```python
# --- Load data ---
data = pd.read_csv("../data/epidemic.csv", parse_dates=["Day"])

# --- Assume known population ---
population = 1_000_000

# --- Build PyMC model ---
model = build_model(data=data, population=population, stochastic=False)  # change to True if needed
```

```python
with model:
    trace = pm.sample(
        draws=1000,
        tune=1000,
        target_accept=0.9,
        return_inferencedata=True,
        random_seed=42,
    )

# --- Save posterior ---
az.to_netcdf(trace, "../results/traces.nc")
```

```python
# --- Posterior plots ---
az.plot_trace(trace, var_names=["beta", "incubation", "symptomatic", "h_rate", "icu_rate", "mu"])
plt.tight_layout()
plt.show()
```
