```python
import daft

def create_epidemiological_model():
    pgm = daft.PGM([8, 5], origin=[0.5, 0.5], grid_unit=2.5)

    # Add nodes with only the short variable name
    pgm.add_node(daft.Node("S", r"$S$", 1, 4))
    pgm.add_node(daft.Node("E", r"$E$", 3, 4))
    pgm.add_node(daft.Node("I", r"$I$", 5, 4))
    pgm.add_node(daft.Node("H", r"$H$", 5, 2))
    pgm.add_node(daft.Node("R", r"$R$", 7, 4))
    pgm.add_node(daft.Node("D", r"$D$", 7, 2))

    # Add edges with horizontal label offsets
    # Moving the labels slightly to the right/left
    pgm.add_edge("S", "E", label=r"$\beta$", label_offset=0.5, label_xoffset=0.5)  # horizontal edge to right
    pgm.add_edge("E", "I", label=r"$\sigma$", label_offset=0.5, label_xoffset=0.5)  # horizontal edge to right
    pgm.add_edge("I", "R", label=r"$\gamma$", label_offset=0.5, label_xoffset=0.5)  # horizontal edge to right
    pgm.add_edge("I", "H", label=r"$h$", label_offset=0.5, label_xoffset=0.5)      # horizontal edge to right

    # Adjust manually for vertical edges (with proper offset horizontally)
    pgm.add_edge("H", "R", label=r"$\gamma_H$", label_offset=0.8, label_angle=90, label_xoffset=0.8)  # vertical edge
    pgm.add_edge("H", "D", label=r"$\mu$", label_offset=0.8, label_angle=90, label_xoffset=0.8)      # vertical edge

    # Render the graph
    pgm.render()
    pgm.figure.savefig("epidemiological_model_with_horizontal_labels.png", dpi=300, bbox_inches='tight')
    print("Saved as 'epidemiological_model_with_horizontal_labels.png'")

create_epidemiological_model()
```
