```python
import daft

def create_epidemiological_model():
    # Create a new Daft plot
    pgm = daft.PGM([6.5, 4.5], origin=[0.5, 0.5], grid_unit=3)

    # Add nodes
    pgm.add_node(daft.Node("S", r"Susceptible", 1, 3, aspect=2.5))
    pgm.add_node(daft.Node("E", r"Exposed", 2.5, 4, aspect=2.5))
    pgm.add_node(daft.Node("I", r"Infected", 4, 3, aspect=2.5))
    pgm.add_node(daft.Node("H", r"Hospitalized", 4, 1.5, aspect=2.5))
    pgm.add_node(daft.Node("R", r"Recovered", 5.5, 4, aspect=2.5))
    pgm.add_node(daft.Node("D", r"Deceased", 5.5, 1, aspect=2.5))

    # Add edges
    pgm.add_edge("S", "E", label=r"Exposure ($\beta$)")
    pgm.add_edge("E", "I", label=r"Infection ($\sigma$)")
    pgm.add_edge("I", "H", label=r"Hospitalization ($h$)")
    pgm.add_edge("I", "R", label=r"Recovery ($\gamma$)")
    pgm.add_edge("H", "R", label=r"Recovery ($\gamma$)")
    pgm.add_edge("H", "D", label=r"Death ($\mu$)")

    # Render and save the plot
    pgm.render()
    pgm.figure.savefig("epidemiological_model_daft.png", dpi=150)
    print("Graphical model visualization saved as 'epidemiological_model_daft.png'")

# Create and save the epidemiological model visualization
create_epidemiological_model()
```
