```python
import daft

def draw_epidemiological_model():
    # Create a new PGM
    pgm = daft.PGM([6.5, 4.5], origin=[0.5, 0.5], grid_unit=2.5, node_unit=1.2, observed_style="inner")

    # Add nodes
    pgm.add_node(daft.Node("S", r"Susceptible", 1, 4))
    pgm.add_node(daft.Node("E", r"Exposed", 2, 4))
    pgm.add_node(daft.Node("I", r"Infected", 3, 4))
    pgm.add_node(daft.Node("H", r"Hospitalized", 4, 4))
    pgm.add_node(daft.Node("R", r"Recovered", 5, 4))
    pgm.add_node(daft.Node("D", r"Deceased", 6, 4))

    # Add edges
    pgm.add_edge("S", "E")
    pgm.add_edge("E", "I")
    pgm.add_edge("I", "H")
    pgm.add_edge("I", "R")
    pgm.add_edge("H", "R")
    pgm.add_edge("H", "D")

    # Render and save the PGM
    pgm.render()
    pgm.figure.savefig("epidemiological_model.png")
    print("Graphical model visualization saved as 'epidemiological_model.png'")

# Draw the epidemiological model
draw_epidemiological_model()
```
